```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("bhanupratapbiswas/superstore-sales")

print("Path to dataset files:", path)
```

```python
import os

files = os.listdir(path)
print(files)
```

```python
import pandas as pd
import os

# Create full file path
file_path = os.path.join(path, 'superstore_final_dataset (1).csv')

# Load dataset with different encoding
df = pd.read_csv(file_path, encoding='latin1')

# Show first 5 rows
df.head()
```

```python
# Display dataset information
df.info()
```

```python
# Summary statistics for numerical columns
df.describe()
```

```python
# Mean
print("MEAN:")
print(df.mean(numeric_only=True))

# Median
print("\nMEDIAN:")
print(df.median(numeric_only=True))

# Mode
print("\nMODE:")
print(df.mode().iloc[0])
```

```python
# Check missing values
print(df.isnull().sum())
```

```python
import matplotlib.pyplot as plt

# Histogram of Sales
plt.hist(df['Sales'], bins=20)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
# Total sales by category
category_sales = df.groupby('Category')['Sales'].sum()

# Bar chart
category_sales.plot(kind='bar')

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
# Count of products by category
category_count = df['Category'].value_counts()

# Pie chart
plt.pie(category_count, labels=category_count.index, autopct='%1.1f%%')

plt.title("Category Distribution")

plt.show()
```
