```python
import pandas as pd
import os

# Create file path
file_path = os.path.join(path, 'traffic.csv')

# Load dataset
df = pd.read_csv(file_path)

# Show first 5 rows
df.head()
```

```python
# Display dataset information
df.info()
```

```python
# Summary statistics
df.describe()
```

```python
# Mean
print("MEAN:")
print(df.mean(numeric_only=True))

# Median
print("\nMEDIAN:")
print(df.median(numeric_only=True))

# Mode
print("\nMODE:")
print(df.mode().iloc[0])
```

```python
# Missing values in each column
print(df.isnull().sum())
```

```python
# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Show all column names
print(df.columns)
```

```python
import matplotlib.pyplot as plt

# Top 10 countries by count
top_countries = df['country'].value_counts().head(10)

# Bar chart
top_countries.plot(kind='bar')

plt.title("Top 10 Countries")
plt.xlabel("Country")
plt.ylabel("Count")

plt.show()
```

```python
# Top 5 events distribution
top_events = df['event'].value_counts().head(5)

# Pie chart
plt.pie(top_events, labels=top_events.index, autopct='%1.1f%%')

plt.title("Top 5 Event Distribution")

plt.show()
```

```python
# Convert date column to datetime
df['date'] = pd.to_datetime(df['date'])

# Count records by date
daily_activity = df.groupby('date').size()

# Line chart
daily_activity.plot(kind='line')

plt.title("Daily Activity Trend")
plt.xlabel("Date")
plt.ylabel("Number of Records")

plt.show()
```
